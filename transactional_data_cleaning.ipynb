- High CLTV customers are the most valuable customer segment.
- They have the highest purchase frequency and average order value.
- Customer retention strategies should focus on loyalty benefits, personalized offers, and repeat-purchase campaigns.

### 2. Medium CLTV Customers
- Medium CLTV customers represent an important growth opportunity.
- Increasing their purchase frequency and order value can help move them toward the High CLTV segment.
- Targeted offers and personalized promotions can encourage repeat purchases.

### 3. Low CLTV Customers
- Low CLTV customers have the lowest purchase frequency and average order value.
- Re-engagement campaigns, targeted discounts, and introductory offers can encourage additional purchases.

### 4. Overall Finding
- High CLTV customers contribute the largest share of total revenue.
- Higher purchase frequency and higher order value are associated with higher CLTV.
- Customer retention should prioritize High CLTV customers, while Medium CLTV customers should be targeted for growth.
