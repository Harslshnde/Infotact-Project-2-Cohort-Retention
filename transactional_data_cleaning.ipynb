```python
import pandas as pd

df = pd.read_excel("../Data/Customer_Transactions_Dataset.xlsx")
```

```python
df.head()
```

```python
total_purchases = (
    df.groupby("Customer_ID")["Transaction_ID"]
    .nunique()
    .reset_index(name="Total_Purchases")
)

total_purchases.head(10)
```

```python
total_revenue = (
    df.groupby("Customer_ID")["Total_Amount"]
    .sum()
    .reset_index(name="Total_Revenue")
)

total_revenue.head(10)
```

```python
aov = (
    total_revenue.merge(total_purchases, on="Customer_ID")
)

aov["Average_Order_Value"] = (
    aov["Total_Revenue"] / aov["Total_Purchases"]
)

aov.head(10)
```

```python
purchase_frequency = (
    df.groupby("Customer_ID")["Transaction_ID"]
    .nunique()
    .reset_index(name="Purchase_Frequency")
)

purchase_frequency.head(10)
```

```python
customer_lifetime = (
    df.groupby("Customer_ID")["Transaction_Date"]
    .agg(["min", "max"])
    .reset_index()
)

customer_lifetime["Customer_Lifetime_Days"] = (
    customer_lifetime["max"] - customer_lifetime["min"]
).dt.days

customer_lifetime.head(10)
```

```python
df["Transaction_Date"] = pd.to_datetime(df["Transaction_Date"])
```

```python
customer_lifetime = (
    df.groupby("Customer_ID")["Transaction_Date"]
    .agg(["min", "max"])
    .reset_index()
)
```

```python
customer_lifetime = (
    df.groupby("Customer_ID")["Transaction_Date"]
    .agg(["min", "max"])
    .reset_index()
)

customer_lifetime["Customer_Lifetime_Days"] = (
    customer_lifetime["max"] - customer_lifetime["min"]
).dt.days

customer_lifetime.head(10)
```

```python
cltv = (
    aov[["Customer_ID", "Total_Revenue", "Total_Purchases", "Average_Order_Value"]]
    .merge(purchase_frequency, on="Customer_ID")
    .merge(
        customer_lifetime[["Customer_ID", "Customer_Lifetime_Days"]],
        on="Customer_ID"
    )
)

cltv.head(10)
```

```python
cltv["CLTV"] = (
    cltv["Average_Order_Value"]
    * cltv["Purchase_Frequency"]
    * (cltv["Customer_Lifetime_Days"] / 365)
)
```

```python
cltv.head(10)
```

```python
top_cltv = (
    cltv.sort_values("CLTV", ascending=False)
    .head(10)
)

top_cltv
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.bar(
    top_cltv["Customer_ID"],
    top_cltv["CLTV"]
)

plt.title("Top 10 Customers by CLTV")
plt.xlabel("Customer ID")
plt.ylabel("CLTV")
plt.xticks(rotation=45)

plt.show()
```
